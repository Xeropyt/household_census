**2.4 Assumptions for Data Preparation**

1. Only renting households are included in the analysis; the study focuses on households living in rented occupied private dwellings. Households that own their homes are excluded because they do not belong to the population being studied. National totals, unclassified household types and special geographic categories are kept separate from the household and area groups being compared. This means the findings describe renting households in selected groups, not all households in Australia or New Zealand.
2. The downloaded NZ rent table combines couple-only families, couples with children and one-parent families into the broader category **One-family household**. Similarly, Australia categories are combined into a similar broad one-family group. However, the countries' definitions are not identcal, so these should be treated as broad comparison groups. Findings should not suggest that every household is classified in exactly the same way in both countries.
3. Categories such as **Not stated**, **Not applicable** and **Not elsewhere included** are excluded when calculating the distribution of rent. Values such as **c..** are treated as missing, never as zero.
4. Totals and their component categories must not be counted together. A total already includes its component categories, for example, adding a **Total households** count to the individual household type counts would count the same households twice. However, small differences between sum of individual categories and a published total can occur due to ABS protecting confidentiality and Stats NZ applying roundings. For the analysis, percentages use the sum of selected classified rent bands consistently, and the counts are not changed to force agreement with published totals. But any large discrepacy is still and should be investigated.
5. Rent bands limit what can be calculated precisely. The exported data give counts within rent ranges, rather than each household's specific weekly rent. An exact median cannot be determined, but later work can identify a median band or clearly label an estimated median. For example, the percentage paying atleast $400 per week can be calculated without estimating individual rents because $400 is a boundary in both datasets. In addition, AUD and NZD amounts are not directly comparable, therefore results should clearly describe clear country, currency and year labels.
6. Australian remoteness categories and NZ regions represent different geographic divisions. Australia's remoteness categories describe relative geographic access to services, including Major Cities, Inner Regional, Outer Regional, Remote and Very Remote areas. NZ regional councils cover regions that can contain both urban and rural communities. Differences in the number, size and composition of these areas also affect the observed gaps.
7. Weekly rent measures housing cost, not affordability. It does not show how easily the household can afford that payment. Two households paying the same rent may experience very different financial pressure because their incomes differ. Assessing affordability requires rent to be considered along with income.
