**3.3 Helper Functions**

These three functions are shared by the cleaning steps below, so each operation is written once and applied the same way wherever it is needed.

`parse_rent_band` turns a rent label from either country into numeric lower and upper boundaries. The upper boundary is exclusive, so adjacent bands meet without gaps, and an open-ended top band is given an upper boundary of infinity.

`add_band_edges` applies that conversion to a whole table, adding the boundaries as new columns while keeping each country's original band labels unchanged.

`sum_remoteness_classes` sums the Australian state-level rows into the five national remoteness classes, after checking that no area code is duplicated and no required count is missing or negative.

```python
# Convert a rent label to numeric boundaries; the upper boundary is exclusive.
# For example, $150-$199 becomes (150, 200), and an open top band ends at infinity.
def parse_rent_band(label):
    text = label.replace("$", "").replace(",", "").lower()
    if text.startswith("under"):
        return 0, int(text.split()[1])
    if "over" in text:
        return int(text.split()[0]), np.inf
    low, high = text.split("-")
    return int(low), int(high) + 1

# Add boundaries to a long table of counts without changing its original rent-band labels.
def add_band_edges(data):
    result = data.copy()
    edges = result["rent_band"].map(parse_rent_band)
    result["band_low"] = edges.str[0]
    result["band_high"] = edges.str[1]
    return result[["country", "area", "rent_band", "band_low", "band_high", "count"]]

# Select ordinary Australian remoteness classes and sum counts across states/territories.
# Required missing cells cause a stop before groupby() can silently skip them.
def sum_remoteness_classes(raw, count_columns):
    assert not raw["RA_CODE_2021"].duplicated().any(), "Duplicate ABS area code."
    data = raw.copy()
    data["area"] = data["RA_CODE_2021"].str[-1].map(RA_LABELS)
    data = data[data["area"].notna()].copy()
    data[count_columns] = data[count_columns].apply(pd.to_numeric)
    assert data[count_columns].notna().all().all(), "Missing required ABS count."
    assert (data[count_columns] >= 0).all().all(), "Negative ABS count."
    return data.groupby("area")[count_columns].sum().reindex(RA_LABELS.values())
```

**3.5 Estimating Median Rent from Banded Data**

RQ3 uses these functions to estimate median weekly rent for each area. Because rent is published in bands, `grouped_median` estimates a median by finding the band that contains the middle household and interpolating linearly within it, assuming households are spread evenly across that band. It returns a missing value if that band is open-ended. `median_table` applies it to each group the caller specifies, such as each country and area. Medians are never averaged across areas, and all medians are reported as estimates.

```python
# Estimate a median by locating the band containing half the cumulative count.
def grouped_median(group, count_col="count"):
    bands = group.groupby(["band_low", "band_high"], as_index=False)[count_col].sum()
    bands = bands.sort_values("band_low")
    counts = bands[count_col].to_numpy(float)
    assert counts.sum() > 0, "A median needs a positive total."
    cumulative = counts.cumsum()
    half = cumulative[-1] / 2
    k = np.searchsorted(cumulative, half)
    below = cumulative[k - 1] if k > 0 else 0.0
    low, high = bands["band_low"].iloc[k], bands["band_high"].iloc[k]
    if np.isinf(high):
        return np.nan  # No finite interpolation is possible in an open-ended band.
    return low + (half - below) / counts[k] * (high - low)

# Apply the median helper to groups selected by the caller, such as country and area.
def median_table(data, by):
    rows = []
    for key, group in data.groupby(by):
        labels = key if isinstance(key, tuple) else (key,)
        row = dict(zip(by, labels))
        row["median_rent"] = grouped_median(group)
        row["households"] = group["count"].sum()
        rows.append(row)
    return pd.DataFrame(rows)
```

**4.2.2 Research Question 2**:

*How does the percentage of renting households paying under 200 AUD per week vary across Australian remoteness classes in 2021, and under 200 NZD per week across New Zealand regional councils in 2023? Which areas have the highest and lowest percentages, and what is the gap within each country?*

**Under $200** can be counted exactly from both countries' bands, without assigning individual rents or estimating a median.

For each area:

**Percentage paying under 200 = 100 × classified-rent households below 200 ÷ all classified-rent households**

For Australia, we add the four rent bands that are completely below 200 AUD, while for New Zealand we use the **Under $200** count. We do not filter by household type or calculate a median rent. Also, the gap is the highest area percentage minus the lowest area percentage, measured in percentage points (pp), and classified rent means rent was stated in a band.

```python
# Made a copy of the cleaned rent data so the original data is not changed for other sections.
rq2_data = rent.copy()

# Create new column for rent below $200. Keep the count if the rent band is below $200 using where(), otherwise change the count to 0 so it is not included. Zero does not mean missing rent values, it means the rent is above $200.
rq2_data["under_200_count"] = rq2_data["count"].where(
    rq2_data["band_high"] <= RENT_THRESHOLD, 0
)

# Group data by country's area using groupby(), and sum() adds total household count and households paying under $200.
rq2_shares = rq2_data.groupby(["country", "area"], as_index=False)[["count", "under_200_count"]].sum()
rq2_shares["percentage"] = 100 * rq2_shares["under_200_count"] / rq2_shares["count"]

# idxmin() and idxmax() identifies the rows containing lowest and highest percentages.
rq2_rows = []
for country, areas in rq2_shares.groupby("country"):
    lowest = areas.loc[areas["percentage"].idxmin()]
    highest = areas.loc[areas["percentage"].idxmax()]
    rq2_rows.append({
        "Country": country,
        "Lowest area": lowest["area"],
        "Lowest (%)": lowest["percentage"],
        "Highest area": highest["area"],
        "Highest (%)": highest["percentage"],
        "Gap (pp)": highest["percentage"] - lowest["percentage"]
    }) # Saving the results to the empty list, rq2_rows.

rq2_summary = pd.DataFrame(rq2_rows)
display(rq2_summary.round(1)) # Round the percentages to 1 decimal place for better readability on the displayed table
```

```python
# This function creates a bar chart from the area percentages for each country.
def plot_rq2(shares, country, year, currency, colour):
    data = shares[shares["country"] == country].set_index("area")["percentage"] # Keep only the areas from the selected country.

    # Put the areas in a useful order for the graph.
    if country == "Australia":
        data = data.reindex(list(RA_LABELS.values())[::-1]) # Reverse the Australian remoteness categories.
    else:
        data = data.sort_values() # Sort NZ areas from lowest to highest percentage.

    # Make the horizontal bar chart.
    ax = data.plot.barh(figsize=(10, max(3.8, 0.38 * len(data))), color=colour)

    # Label the percentage value beside each bar.
    ax.bar_label(ax.containers[0], fmt="%.1f%%", padding=4)

    # Title, axis labels and percentage range.
    ax.set_title(f"{country} ({year}): weekly rent under {currency} {RENT_THRESHOLD}")
    ax.set_xlabel("Percentage of households with classified rent (%)")
    ax.set_ylabel("Remoteness class" if country == "Australia" else "Regional council")
    ax.set_xlim(0, 100)
    plt.tight_layout() # Leave room for the area names and labels.
    plt.show()

plot_rq2(rq2_shares, "Australia", 2021, "AUD", "#287A9F") # AUS graph
plot_rq2(rq2_shares, "New Zealand", 2023, "NZD", "#A9583D") # NZ graph
```

RQ2 finds a clear remoteness pattern in Australia and a less simple regional pattern in NZ. These patterns however does not establish that location causes rent differences. **4.2.3 Research Question 3**:

*Do areas with a higher share of a given household type also have a higher estimated median weekly rent, and is this relationship consistent between Australia and New Zealand or does it diverge?*

```python
# Household-type shares by area (built in 3.4.3 and 3.4.4), so RQ3 runs without RQ1
rq3_shares = pd.concat({"Australia": au_household_percentages[HOUSEHOLD_TYPES],
                        "New Zealand": nz_household_percentages[HOUSEHOLD_TYPES]},
                       names=["country", "area"])

# Estimated median rent for each area (classified rent only), joined to the shares
area_rent = median_table(rent, ["country", "area"]).set_index(["country", "area"])
rq3 = rq3_shares.join(area_rent["median_rent"])
assert rq3["median_rent"].notna().all(), "An area's median falls in the open top band."
display(rq3.round(1))

# Correlation between each household type's share and median rent, within each country
rq3_corr = rq3.groupby(level="country").apply(
    lambda d: d[HOUSEHOLD_TYPES].corrwith(d["median_rent"]))
display(rq3_corr.round(2))

fig, axes = plt.subplots(1, len(HOUSEHOLD_TYPES), figsize=(14, 4))
for ax, hh_type in zip(axes, HOUSEHOLD_TYPES):
    for country, d in rq3.groupby(level="country"):
        ax.scatter(d[hh_type], d["median_rent"], label=country)
    ax.set_title(hh_type)
    ax.set_xlabel("Share of households (%)")
axes[0].set_ylabel("Estimated median weekly rent (local currency)")
axes[0].legend()
plt.tight_layout()
plt.show()

# Stability check: NZ correlations with the most extreme region removed
nz_rq3 = rq3.loc["New Zealand"]
for region in ["West Coast Region", "Auckland Region"]:
    d = nz_rq3.drop(region)
    print(f"Without {region}:")
    print(d[HOUSEHOLD_TYPES].corrwith(d["median_rent"]).round(2), "\n")
```

**Analysis:**

**Australia (2021):** Across the five remoteness classes, areas with a higher share of lone-person households have a lower estimated median weekly rent (r = −0.88), while family (r = 0.82) and group (r = 0.85) shares are positively associated with rent. Major Cities of Australia has the highest family (71.1%) and group (4.2%) shares, the lowest lone-person share (24.6%), and the highest median rent (about AUD 410). Very Remote Australia has the lowest median (about AUD 132). However, household composition varies very little across remoteness classes: family shares range only from 67.7% to 71.1%, and group shares from 2.7% to 3.3% outside Major Cities. Rent, by contrast, roughly triples between Very Remote and Major Cities. The strong correlations are therefore driven mainly by Major Cities differing from all other classes, and they rest on only five data points.

**New Zealand (2023):** The direction of the relationship is the same. Lone-person share is negatively associated with median rent (r = −0.76), and family (r = 0.53) and group (r = 0.34) shares are positively associated. Auckland has the highest family share (71.0%) and the highest median rent (about NZD 545), while West Coast has by far the highest lone-person share (44.4%) and the lowest median rent (about NZD 260). The lone-person relationship is robust: it stays at about −0.70 when either West Coast or Auckland is removed. The family relationship is less stable, falling to 0.29 without West Coast, so it depends heavily on that one region. Group households show the weakest link. Otago has the highest group share (19.2%) but a mid-range median rent (about NZD 413), while Wellington combines a high group share (14.8%) with the second-highest rent (about NZD 508).

**Answer to RQ3:** In both countries, areas with more family and group households tend to have higher median rents, and areas with more lone-person households tend to have lower rents. The relationship is consistent in direction between Australia and New Zealand. It differs in strength: Australia's correlations are stronger but come from only five areas with small compositional differences, while New Zealand's 16 regions show a clear negative link for lone-person households and weaker, less stable positive links for family and group households.

These are area-level associations. They don't show that any particular household type pays more rent, and both household mix and rent are likely shaped by a shared factor such as urbanisation. The Australian shares also cover all tenure types, while the NZ shares cover renting households only, and AUD and NZD values are not directly comparable.
